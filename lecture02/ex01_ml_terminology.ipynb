# Machine Learning Terminology

This notebook introduces the core vocabulary of machine learning -- instances, attributes, features, target, model, parameters and hyperparameters -- using the Iris dataset as a running example.

## Import Required Libraries

Import numpy, pandas and the scikit-learn tools used below.

```python
# Import the required libraries.
import numpy as np
import pandas as pd

from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
```

## Load the Iris Dataset

Load the classic Iris dataset, a bundle of features, target and descriptive metadata.

```python
# Load the Iris dataset as a scikit-learn Bunch.
iris = load_iris()

# Show the available keys of the dataset object.
print(f"Keys: {list(iris.keys())}")
```

## Instances and Attributes

Each row is an instance and each column an attribute; the feature matrix shape reveals how many of each we have.

```python
# Inspect the shape and dtype of the feature matrix.
X = iris.data
y = iris.target
print(f"Feature matrix shape (instances, attributes): {X.shape}")
print(f"Feature dtype: {X.dtype}")
print(f"Target shape: {y.shape}, dtype: {y.dtype}")
```

## Feature and Class Names

The four features describe flower measurements and the target encodes three iris species.

```python
# Show the feature names and the class names.
print(f"Features: {list(iris.feature_names)}")
print(f"Classes: {list(iris.target_names)}")
```

## The Data Matrix as a DataFrame

Assemble the features and the target into a single DataFrame to see the data matrix the way a model consumes it.

```python
# Build a DataFrame with features and the encoded target.
df = pd.DataFrame(X, columns=iris.feature_names)
df["target"] = y
df["species"] = df["target"].map(
    dict(enumerate(iris.target_names)))
df.head()
```

## Identifying the Target Column

The `target` column is the label to predict; every other numeric column is an input feature.

```python
# Point out which column is the target and its distribution.
feature_cols = list(iris.feature_names)
target_col = "target"
print(f"Input feature columns: {feature_cols}")
print(f"Target column: {target_col}")
print("Class counts:")
print(df["species"].value_counts())
```

## Model, Parameters and Hyperparameters

Fitting a model learns parameters from data, while hyperparameters are configuration values we set beforehand.

```python
# Fit a model and separate learned parameters from hyperparameters.
model = LogisticRegression(max_iter=200, C=1.0)
model.fit(X, y)

# Learned parameters end with an underscore.
print(f"Learned coefficients shape: {model.coef_.shape}")
print(f"Learned intercepts: {model.intercept_.round(2)}")

# Hyperparameters are the values we chose before fitting.
print(f"Hyperparameter C: {model.C}")
print(f"Hyperparameter max_iter: {model.max_iter}")
```

## Summary

- **Instance**: a single row of the data matrix (one iris flower).
- **Attribute / feature**: a column used as model input.
- **Target**: the column to predict, never used as an input.
- **Parameters**: values learned during `fit` (end with `_`).
- **Hyperparameters**: values we set before fitting, such as `C` and `max_iter`.
